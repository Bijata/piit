```python
import numpy as np
from scipy import stats
```

```python
#Hypothesis One sample T-test
Sample_size = 25
Sample_mean_weight = 300
Sample_standard_deviation = 18.5
#H0: μ = 310 (population mean is equal to 310 pounds)
#H1: μ ≠ 310 (population mean is not equal to 310 pounds)
hypothesized_mean = 310  
```

```python
np.random.seed(30)
sample_weights = np.random.normal(loc=Sample_mean_weight, scale=Sample_standard_deviation, size=Sample_size)
sample_weights
```

```python
#data, popmean from eg
t_statistic, p_value = stats.ttest_1samp(a = sample_weights , popmean = hypothesized_mean)
print(t_statistic, p_value)
```

```python
if p_value < 0.05:
    print("Conclusion: Reject the null hypothesis", p_value)
    print("The sample mean is significantly different from the hypothesized population mean.")
else:
    print("Conclusion: Fail to reject the null hypothesis.")
    print("There is no significant difference between the sample mean and the hypothesized population mean.")
```

```python
# Generate a sample of 20 widget weights
np.random.seed(42) # for reproducibility
sample_weights = np.random.normal(loc=51.5, scale=2, size=20)
print("Sample weights:", np.round(sample_weights, 2))
```

```python
# Define the hypothesized population mean
pop_mean = 50

# Perform the one-sample t-test
t_statistic, p_value = stats.ttest_1samp(a=sample_weights, popmean=pop_mean)

print(f"T-statistic: {t_statistic:.4f}")
print(f"P-value: {p_value:.4f}")
```

```python
Conclusion : P-value < 0.05 reject H0 (no effect)
```
